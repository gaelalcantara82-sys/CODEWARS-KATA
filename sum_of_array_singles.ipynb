## Sum of array singles

```python
def repeats(arr):
    resultado = 0
    
    for n in set(arr):
        contador= arr.count(n)
        
        if contador == 1:
            resultado += n
    
    return resultado
            
```

2. Hago una variable llamada contador que cuenta el numero de veces que aparece un numero del array.

3. Si el contador solo cuenta una vez el número lo suma a la variable resultado.
