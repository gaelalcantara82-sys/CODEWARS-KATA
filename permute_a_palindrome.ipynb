## Permute a palindrome

```python
def permute_a_palindrome(input):
    impares = 0
    
    for char in set(input):
        contador= input.count(char)
    
        if contador % 2 != 0:
            impares += 1
        
    if impares <= 1:
        return True
    else:
        return False
```

1. Creo la variable impares y le doy valor 0.

2. Creo una variable llamada contador que tendrá como valor el número de veces que aparezca una letra.

3. Si el numero al dividirlo entre dos tiene un resto distinto a 0, el número es impar, y a la variable se le suma 1.

4. Si el número de impares es mayor a 1, nunca vamos a poder reorganizarlo para que sea un palindromo, por lo que si el número es menor o igual a 1, tiene que devolver verdadero, y si impares es mayor a 1, tiene que devolver falso.
